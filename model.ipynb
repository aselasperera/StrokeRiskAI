# Part A: Application Area Review – AI for Stroke Prediction in Healthcare

## Introduction

The second cause of death in the whole world is stroke. Approximately 15 million individuals experience strokes each year all over the world. Five million of these succumb and the rest are permanently disabled. Early screening of individuals who have high risks of stroke is quite essential as this enables the treating physicians to offer preventive care such as medication, lifestyle modifications and more frequent observations.

Risk of stroke can be predicted with the help of artificial intelligence and machine learning through the examination of patient information. Such AI systems are capable of handling complex data such as age, medical history, blood pressure and lifestyle to provide an individual with risk scores that assist doctors in making improved decisions.

## AI Techniques in Stroke Prediction

**Logistic Regression** is a conventional statistical tool that has been employed in medicine over the years. It determines the risk of stroke by establishing a combination of various risk factors with a particular weight. This is why doctors use this approach as they can visualize the contributions made by each factor (such as age or blood pressure) to the ultimate risk score. The literature indicates that logistic regression has an accuracy of 70-75% in prediction of strokes. The key limitation lies in the assumption of the simple relationships between the risk factors and the stroke that can be too simplistic to represent the complex interactions.

**Neural Networks** The more advanced AI models are the Neural Networks that have the capability to learn complicated patters automatically. These are models that apply several layers of calculations in order to find concealed relationships in the data. It has been demonstrated that neural networks are capable of predicting stroke with 80-85% accuracy after learning the interaction between various risk factors. Its biggest disadvantage is that neural networks are black boxes - physicians cannot simply learn why this or that prediction was made by the model.

**Random Forest** is an intermediate strategy which constructs several decision trees and averages their results. Research indicates that the Random Forest has attained 75-82 percent accuracy in predicting stroke, as well as giving information on the risk factors that are the most significant. The approach is effective when working with medical data and has a fair balance between interpretation and accuracy. 

## Clinical Applications and Challenges

It seems that AI stroke prediction models could be employed during the screening of patients in the clinic, making decisions based on whether the patient requires preventive treatment, prioritizing urgent cases in the emergency room and managing population health programs.

However, it has significant challenges. The quality of data is different in various hospitals. Biases are of concern - models may not perform equally with the different age groups, sexes, or races. The legal principles of application of AI in medical decision-making are under development.

## Future Directions

In the future, they are going to work with several forms of data (clinical records, images, genetics, wearable devices) in a privacy-friendly manner, and create explainable AI. With the advancement of such technologies, stroke prediction with the help of AI might become a routine preventive measure in healthcare.

# Part B: Comparison of AI Techniques for Stroke Prediction

This part is an assessment of three machine learning strategies to predict strokes using actual clinical information. One of the techniques will be chosen after the comparison and will be applied in Part C.

## Technique 1: Logistic Regression

**How it Works:**  
Logistic Regression is a mathematical method of determining the likelihood of stroke using input features. It trains weights of every feature (age, blood pressure, etc.) and uses them to predict the risk of stroke.

**Application to Stroke Dataset:**  
In the case of healthcare stroke data (demographic (age, gender), clinical (blood pressure, BMI, glucose), and lifestyle (smoking, work type) variables, Logistic Regression would be used to learn the contribution of all variables to the risk of stroke on an individual basis.

**Strengths:**
- Very easy to interpret - every weight indicates the precise importance of every factor.
- High-speed training and forecasting.
- Navy veteran in medicine.
- Even smaller datasets can be used.

**Weaknesses:**
- Unable to acquire complicated associations among features automatically.
- Makes assumptions of simple linear relationships.
- Relationships are complex and May misses out on essential patterns.

**Expected Performance:** 
- The performance is expected to be 70-75 percent.

---

## Technique 2: Neural Network (Multi-Layer Perceptron)

**How it Works:**  
Nearly any pattern can be learned through a Neural Network that incorporates several layers of computation that make use of non-linear functions. It works with data on hidden layers that automatically identify significant combinations of features.

**Application to Stroke Dataset:**  
All the features would be processed by the neural network through the hidden layers (64-128 neurons per layer) to learn the intermediate representations. As an example, it may automatically inquire, age over 65 and high blood pressure and high glucose are a very high risk.

**Strengths:**
- Is able to learn non-linear complex patterns automatically.
- Higher accuracy potential (78-85%)
- Can find feature interactions through automatic programming.
- Agile with numerous features.

**Weaknesses:**
- Black box - not very easily comprehended how it was able to give certain predictions.
- Inefficiently sensitive to a variety of parameters.
- Takes longer to train
- Requires additional data to be at its best.
- May overfits in the untypical methods.

**Expected Performance:** 78-85% performance predicted.

---

## Technique 3: Random Forest

**How it Works:**  
Random Forest constructs numerous decision trees of various subsets of data and sums up their votes. Every tree is trained on different patterns and majority vote is taken as the final prediction.

**Application to Stroke Dataset:**  
Random Forest would create 100-500 trees and each tree would define its rules such as IF age >65 and hypertension=1 THEN high risk. The model will offer the scores of the importance of the features indicating which variables are the most important.

**Strengths:**
- Medical data tabular improvement (75-82%) is excellent.
- Offers importance ranking of features.
- Naturally supports the various types of data.
- Less prone to overfitting
- Requires minimal tuning

**Weaknesses:**
- Larger model size (stores many trees)
- Slower predictions than Logistic Regression
- Less interpretable than Logistic Regression

**Expected Performance:** 75-82% accuracy

## C.3: Data Preprocessing and Feature Engineering

```python
print("="*60)
print("DATA PREPROCESSING")
print("="*60)

# Create working copy
df_processed = df.copy()
print(f"\nStarting with {df_processed.shape[0]:,} rows and {df_processed.shape[1]} columns")

# Step 1: Drop ID column
print("\n[1/7] Dropping ID column...")
df_processed = df_processed.drop('id', axis=1)
print("    ✓ Removed: id")

# Step 2: Handle BMI missing values
print("\n[2/7] Handling BMI missing values...")
# Convert 'N/A' strings to NaN
df_processed['bmi'] = pd.to_numeric(df_processed['bmi'], errors='coerce')
missing_bmi = df_processed['bmi'].isnull().sum()
print(f"    Missing BMI values: {missing_bmi} ({missing_bmi/len(df_processed)*100:.1f}%)")

# Impute with median
bmi_median = df_processed['bmi'].median()
df_processed['bmi'] = df_processed['bmi'].fillna(bmi_median)
print(f"    ✓ Imputed with median: {bmi_median:.2f}")
print(f"    Missing after imputation: {df_processed['bmi'].isnull().sum()}")
```

```python
# Step 3: Feature engineering - create risk indicators
print("\n[3/7] Creating clinical risk features...")

# Age risk categories
df_processed['age_risk_high'] = (df_processed['age'] >= 65).astype(int)
df_processed['age_risk_moderate'] = ((df_processed['age'] >= 50) & (df_processed['age'] < 65)).astype(int)

# BMI categories
df_processed['obese'] = (df_processed['bmi'] >= 30).astype(int)
df_processed['overweight'] = ((df_processed['bmi'] >= 25) & (df_processed['bmi'] < 30)).astype(int)

# Glucose categories
df_processed['high_glucose'] = (df_processed['avg_glucose_level'] >= 140).astype(int)
df_processed['prediabetic'] = ((df_processed['avg_glucose_level'] >= 100) & 
                                (df_processed['avg_glucose_level'] < 140)).astype(int)

# Cardiovascular risk score
df_processed['cv_risk_score'] = (
    df_processed['hypertension'] + 
    df_processed['heart_disease'] + 
    df_processed['age_risk_high']
)

# Metabolic risk (combination of obesity and high glucose)
df_processed['metabolic_risk'] = ((df_processed['obese'] == 1) & 
                                   (df_processed['high_glucose'] == 1)).astype(int)

# Interaction features
df_processed['age_x_hypertension'] = df_processed['age'] * df_processed['hypertension']
df_processed['age_x_heart_disease'] = df_processed['age'] * df_processed['heart_disease']
df_processed['bmi_x_glucose'] = df_processed['bmi'] * df_processed['avg_glucose_level'] / 100

print("    ✓ Created 11 engineered features:")
print("      - Age risk categories (high, moderate)")
print("      - BMI categories (obese, overweight)")
print("      - Glucose categories (high, prediabetic)")
print("      - CV risk score")
print("      - Metabolic risk indicator")
print("      - 3 interaction features")
```

```python
# Step 4: Encode target variable
print("\n[4/7] Encoding target variable...")
y = df_processed['stroke'].values
df_processed = df_processed.drop('stroke', axis=1)
print(f"    ✓ Target variable extracted")
print(f"    Class 0 (No Stroke): {(y==0).sum():,}")
print(f"    Class 1 (Stroke):    {(y==1).sum():,}")
```

```python
# Step 5: Encode categorical variables
print("\n[5/7] Encoding categorical variables...")

# Identify categorical columns
categorical_cols = ['gender', 'ever_married', 'work_type', 'Residence_type', 'smoking_status']
print(f"    Categorical columns: {categorical_cols}")

# One-hot encode
df_encoded = pd.get_dummies(df_processed, columns=categorical_cols, drop_first=True)
print(f"    ✓ One-hot encoded {len(categorical_cols)} categorical features")
print(f"    Shape after encoding: {df_encoded.shape}")
```

```python
# Step 6: Scale numerical features
print("\n[6/7] Scaling numerical features...")

# Identify numerical columns (will be all columns after one-hot encoding)
X = df_encoded.values
feature_names = df_encoded.columns.tolist()

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print(f"    ✓ Scaled {X_scaled.shape[1]} features using StandardScaler")
print(f"    Features now have mean≈0, std≈1")
```

```python
# Step 7: Train-test split BEFORE handling imbalance
print("\n[7/7] Train-test split...")

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, stratify=y, random_state=42
)

print(f"    Training set: {len(X_train):,} samples")
print(f"      - No Stroke: {(y_train==0).sum():,} ({(y_train==0).sum()/len(y_train)*100:.1f}%)")
print(f"      - Stroke:    {(y_train==1).sum():,} ({(y_train==1).sum()/len(y_train)*100:.1f}%)")

print(f"    Test set: {len(X_test):,} samples")
print(f"      - No Stroke: {(y_test==0).sum():,} ({(y_test==0).sum()/len(y_test)*100:.1f}%)")
print(f"      - Stroke:    {(y_test==1).sum():,} ({(y_test==1).sum()/len(y_test)*100:.1f}%)")

print("\n" + "="*60)
print("PREPROCESSING COMPLETE")
print("="*60)
print(f"Final feature count: {X_scaled.shape[1]}")
```

```python
# Handle class imbalance with SMOTE on training data only
print("\n" + "="*60)
print("HANDLING CLASS IMBALANCE WITH SMOTE")
print("="*60)

print("\nBefore SMOTE:")
print(f"  Training samples: {len(y_train):,}")
print(f"  Class 0: {(y_train==0).sum():,}")
print(f"  Class 1: {(y_train==1).sum():,}")
print(f"  Ratio: {(y_train==0).sum()/(y_train==1).sum():.1f}:1")

smote = SMOTE(random_state=42)
X_train_balanced, y_train_balanced = smote.fit_resample(X_train, y_train)

print("\nAfter SMOTE:")
print(f"  Training samples: {len(y_train_balanced):,}")
print(f"  Class 0: {(y_train_balanced==0).sum():,}")
print(f"  Class 1: {(y_train_balanced==1).sum():,}")
print(f"  Ratio: {(y_train_balanced==0).sum()/(y_train_balanced==1).sum():.1f}:1")

print("\n✓ Dataset is now balanced for training!")
print("  (Test set remains unchanged to reflect real-world distribution)")
```

## C.4: Neural Network Training with Hyperparameter Tuning

```python
print("="*60)
print("NEURAL NETWORK (MLP) TRAINING")
print("="*60)

# Define hyperparameter search space
param_grid = {
    'hidden_layer_sizes': [(64,), (128,), (64, 32), (128, 64), (128, 64, 32)],
    'alpha': [0.0001, 0.001, 0.01],
    'learning_rate_init': [0.001, 0.01],
    'batch_size': [32, 64, 128],
    'max_iter': [500]
}

print("\nHyperparameter search space:")
for param, values in param_grid.items():
    print(f"  {param}: {values}")

total_combinations = 1
for values in param_grid.values():
    total_combinations *= len(values)

print(f"\nTotal possible combinations: {total_combinations}")
print(f"RandomizedSearchCV will sample: 15 combinations")
print(f"With 5-fold CV: 15 × 5 = 75 model fits")
```

```python
# Perform RandomizedSearchCV
print("\n" + "="*60)
print("HYPERPARAMETER OPTIMIZATION")
print("="*60)
print("\nThis will take 10-15 minutes...\n")

cv_strategy = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

mlp_search = RandomizedSearchCV(
    MLPClassifier(early_stopping=True, validation_fraction=0.1, random_state=42, 
                  activation='relu', solver='adam'),
    param_grid,
    n_iter=15,
    cv=cv_strategy,
    scoring='roc_auc',
    random_state=42,
    n_jobs=-1,
    verbose=2
)

mlp_search.fit(X_train_balanced, y_train_balanced)

print("\n" + "="*60)
print("OPTIMIZATION COMPLETE")
print("="*60)
```

```python
# Display best parameters
print("\n" + "="*60)
print("BEST HYPERPARAMETERS")
print("="*60)

for param, value in mlp_search.best_params_.items():
    print(f"  {param}: {value}")

print(f"\nBest Cross-Validation ROC-AUC: {mlp_search.best_score_:.4f}")

# Top 5 configurations
print("\n" + "="*60)
print("TOP 5 CONFIGURATIONS")
print("="*60)

results_df = pd.DataFrame(mlp_search.cv_results_)
results_df = results_df.sort_values('rank_test_score')
top5 = results_df[['params', 'mean_test_score', 'std_test_score']].head(5)

for idx, (_, row) in enumerate(top5.iterrows(), 1):
    print(f"\nRank {idx}: ROC-AUC = {row['mean_test_score']:.4f} (+/- {row['std_test_score']:.4f})")
```

```python
# Cross-validation with all metrics
print("\n" + "="*60)
print("CROSS-VALIDATION PERFORMANCE (BEST MODEL)")
print("="*60)

scoring_metrics = {
    'accuracy': 'accuracy',
    'precision': 'precision',
    'recall': 'recall',
    'f1': 'f1',
    'roc_auc': 'roc_auc'
}

cv_results = cross_validate(
    mlp_search.best_estimator_,
    X_train_balanced,
    y_train_balanced,
    cv=cv_strategy,
    scoring=scoring_metrics
)

print("\nMetric          Mean      Std Dev")
print("-" * 40)
for metric in scoring_metrics.keys():
    scores = cv_results[f'test_{metric}']
    print(f"{metric.upper():12s}    {scores.mean():.4f}    (+/- {scores.std():.4f})")
```

## C.5: Final Model Evaluation on Test Set

# Part D: Software Testing and Evaluation

## Testing Methodology

In the testing of the neural network, the machine learning principles were applied to have a reliable performance evaluation of the real medical data.

**Train-Test Split:**  
The 5,110 patient data was stratified into 80% training (4,088 patients) and 20% testing (1,022 patients) datasets with the help of stratification. This preserves the natural 95:5 class distribution in either of the sets - 95 No Stroke, 5 Stroke. The test set was fully reserved and not used in the development to provide objective outcomes.

**Handling Class Imbalance:**  
The major challenge is the severe imbalance (95% No Stroke, 5% Stroke). With no treatment, a model would be able to make 95 percent accuracy by simply classifying everyone as No Stroke, and none of the people with actual strokes. To work around this produced synthetics (SMOTE) was used on training data alone, synthetically producing stroke cases to balance the classes to 50:50. This assists the neural network to learn and identify strokes rather than making an assumption of the majority class. Notably, test data was not subjected to SMOTE - it preserves a 95:5 distribution which is realistic.

**Cross-Validation and Hyperparameter Tuning:**  
The balanced training data were cross-validated 5 times. RandomizedSearchCV was used to test these 15 hyperparameter combinations:

- Hidden layer sizes: (64,), (128,), (64,32), (128,64), (128,64,32)
- Regularization (alpha): 0.0001, 0.001, 0.01
- Learning rates: 0.001, 0.01
- Batch sizes: 32, 64, 128

The best configuration - (128, 64) architecture with alpha=0.01, learning_rate=0.001, batch_size=128 - was selected based on ROC-AUC performance.

**Evaluation Metrics:**  
Multiple metrics were calculated:
- **Accuracy (89.82%):** Overall correctness
- **Precision (10.29%):** Of predictions labeled "Stroke", how many were actually strokes
- **Recall (14.00%):** Of actual stroke cases, how many were caught
- **F1-Score (11.86%):** Balance between precision and recall
- **ROC-AUC (0.7241):** Ability to discriminate between stroke and non-stroke across all thresholds

## Equality, Diversity, and Inclusion Analysis

### Algorithmic Bias and Fairness

The prediction systems of AI strokes should be discussed as they might have certain biases that would increase the disparities in health.

**Age Bias:**  
The model specifically forms an age 65 and above = high risk factor, which might promote discrimination as age. The model may also be excessive in age, and overlook the risk in persons with severe risk factors (such as a 45 year old with severe hypertension and diabetes) who are not old enough. On the contrary, the elder patients may be systematically classified as high-risk due to their age rather than modifiable variables.

The model had only a 7 out of 50 stroke capture. When these 7 people are mostly the elderly and the 43 people who missed the strokes include people of ages, this will be discriminatory. This bias would be identified by age-stratified assessment (performance analysis in various age groups) which was not done.

**Gender Bias:**  
Stroke manifests itself differently between the sexes. The symptoms that are characterized as atypical in women include severe fatigue and nausea, whereas typical symptoms are manifested in men through weakness and speech difficulty. When training data is biased to men patterns then the model could be ineffective to women.

There are also other risk factors in women (pregnancy complications, oral contraceptives, hormonal changes). The model is not able to explain sex-specific factors, which could give biased predictions of gender.

**Socioeconomic Bias:**  
Such characteristics as worktype and Residencetype (Urban/Rural) are crudely used as a proxy of socioeconomic status including important items such as income, education, insurance, and neighborhood resources.

Living in the rural area does not only impact the risk of stroke but also the availability of stroke-capable hospitals and emergency services. The model is not able to differentiate between access barriers and biological risk. It is possible that labeling rural patients as high-risk and lacking preventive services may have negative effects.

**Racial Disparities - Critical Limitation:**  
The Kaggle dataset has no race/ethnicity variables. This is a major problem because:
- Black individuals experience stroke at nearly twice the rate of White individuals
- Strokes occur at younger ages in Black populations (median age 63 vs 72)
- Hispanic individuals have elevated stroke risk
- Asian populations show different stroke patterns

By omitting race, the model cannot account for these differences. This creates potential for systematic underestimation of risk in high-incidence racial groups. A 55-year-old Black patient with moderate hypertension might have much higher stroke risk than a 55-year-old White patient with identical features, but the model treats them the same.
